### INDEXING

#### 1. 데이터 로드 (Load)

```python
import glob
files = glob.glob('docs/*.txt') # docs 폴더에 있는 모든 텍스트 파일 조회
print(files)

# 문서 로드(Load Documents)
from langchain_community.document_loaders import TextLoader # 파일 로딩 설정
docs = []
for file in files:
    docs = docs + TextLoader(file).load() # 로더 설정 + 문서 로드
len(docs) # 로드된 문서 수 확인
```

```python
docs[:2]
```

```python
# 데이터 전처리
for doc in docs:
    doc.page_content = doc.page_content.replace('\x07', ' ').replace('\t', ' ').replace('  ', ' ')
print(docs[4].page_content)
```

#### 2. 텍스트 분할 (Split)

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter # **

# 문자열 분할기 설정
text_splitter = RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=0) # chunk_overlap : 보통 chunk_size의 10~20%

# 문서 분할 : 이미 로드된 문서(docs)를 대상으로 텍스트 분할기를 적용 -> split_documents()
split_docs = text_splitter.split_documents(docs)

# 로드된 문서의 수 확인
print(len(split_docs)) # chunksize 600 -> 1431, chunksize 800 -> 1074

# 다섯번째 문서 확인
split_docs[4]
```

#### 3. 임베딩 (Embed)

```python
# '임베딩은 Cache Embedding 사용을 권장합니다.'

from langchain.storage import InMemoryByteStore
from langchain_openai import OpenAIEmbeddings
from langchain.embeddings import CacheBackedEmbeddings # 속도 up

# OpenAI(표준화)의 "text-embedding-3-small" 모델을 사용하여 임베딩을 생성합니다.(1024 차원)
embeddings = OpenAIEmbeddings(model="text-embedding-3-small", dimensions=1024)

# 인메모리 스토어를 생성합니다.
store = InMemoryByteStore()

# 캐시를 지원하는 임베딩 생성
cached_embeding = CacheBackedEmbeddings.from_bytes_store(
    underlying_embeddings=embeddings,
    document_embedding_cache=store,
    namespace=embeddings.model,  # 기본 임베딩과 저장소를 사용하여 캐시 지원 임베딩을 생성
    query_embedding_cache=True
)
```

```python
# test
# 텍스트를 임베딩하여 쿼리 결과를 생성
query_result = cached_embeding.embed_query("임베딩 테스트를 하기 위한 샘플 문장입니다.")
print(query_result[:5])
print(len(embeddings.embed_documents(['AI Essential'])[0]))

# 임베딩 대상 텍스트
sentence1 = "안녕하세요? 반갑습니다."
sentence2 = "안녕하세요? 반갑습니다!"
sentence3 = "안녕하세요? 만나서 반가워요."
sentence4 = "Hi, nice to meet you."
sentence5 = "I like to eat apples."

# 임베딩 수행
sentences = [sentence1, sentence2, sentence3, sentence4, sentence5]
embedded_sentences = embeddings.embed_documents(sentences) # 각 텍스트를 1024차원으로 바뀜

from sklearn.metrics.pairwise import cosine_similarity

def similarity(a, b):
    return cosine_similarity([a], [b])[0][0]

# 유사도 계산
# 의미가 같으면 높게 나오고, 언어가 동일하면 높게 나옴
for i, sentence in enumerate(embedded_sentences):
    for j, other_sentence in enumerate(embedded_sentences):
        if i < j:
            print(
                f"[유사도 {similarity(sentence, other_sentence):.4f}] {sentences[i]} \t <=====> \t {sentences[j]}"
            )
```

#### 4. 저장 (Store)

```python
# FAISS 벡터 스토어 생성
# FAISS DB
from langchain_community.vectorstores import FAISS

# DB 생성 + 문서 저장
db = FAISS.from_documents(documents=split_docs,
                          embedding=OpenAIEmbeddings(model="text-embedding-3-small"))
```

```python
# test
# 저장된 문서 개수 확인 (문서 전체 정상 저장됨?)
print(len(db.index_to_docstore_id))
# 1074개 : 아까 split 해둔 chunk 개수 만큼! => 전체 정상 저장됨 확인

# 문서 저장소 ID 확인
print(db.index_to_docstore_id)

# 저장된 문서 내용 확인 (위 두개만 직접 비교)
print(db.docstore._dict)

# 유사도 검색
print(db.similarity_search("Transformer"))
# "Transformer"라는 단어를 벡터로 임베딩한 다음, 그 벡터와 비슷한 의미를 가진 문서를 벡터 DB에서 찾기
```

#### 2. RagChain 생성

```python
# Prompt
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    # System messaage
    """당신은 질문-답변(Question-Answering)을 수행하는 친절한 AI 어시스턴트입니다.
    당신의 임무는 주어진 문맥(context) 에서 주어진 질문(question) 에 답하는 것입니다.
    검색된 다음 문맥(context) 을 사용하여 질문(question) 에 답하세요.
    또한 어떤 문맥을 참고했느지 출처를 같이 표현해주세요.
    만약, 주어진 문맥(context) 에서 답을 찾을 수 없다면, 답을 모른다면 주어진 정보에서 질문에 대한 정보를 찾을 수 없습니다 라고 답하세요.
    한글로 답변해 주세요. 단, 기술적인 용어나 이름은 번역하지 않고 그대로 사용해 주세요.

#Question:
{question}

#Context:
{context}

#Answer:"""
)
```

```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name="gpt-4o-mini", temperature=0)
retriever = db.as_retriever()

# 체인을 생성합니다.
rag_chain = (
    {"context": retriever, "question": RunnablePassthrough()} # *** 이게 일반LLM 과 다른 부분
    | prompt
    | llm
    | StrOutputParser()
)
```

#### 3. 답변 생성 (정답 제출)

```python
# test: 답변 생성
# 아는 질문
print(rag_chain.invoke('DataLoader 객체에 대해 알려줘'))

# 모르는 질문
print(rag_chain.invoke('우리 가족 구성원 알려줘'))
```

```python
# 정답 제출
# # [예시]
# # 실제 제출 시에는 Chain의 결과를 바로 전달하시면 됩니다. (제출할때 이부분 풀어서 제출 )
# # 전달할 값은 꼭 문자열이어야 합니다.
for question_id in comp.questions:
    comp.questions[question_id]['answer'] = str(rag_chain.invoke(comp.questions[question_id]['question']))
```
