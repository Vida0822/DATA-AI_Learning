```python
%%writefile app.py
# streamlit 패키지 추가
import streamlit as st

# Title
# :색상이름[내용]으로 색상 지정 가능
st.title('나는 :red[Title]입니다.')

# Header
# divider=True로 임의 색상 분할선 추가
st.header('나는 :blue[Header]입니다.', divider=True)

# Subheader
# divider=색상으로 지정 색상 분할선 추가 (색상은 blue, green orange, red, violet, gray/grey 가능)
st.subheader('나는 :violet[Subheader]입니다.', divider='violet')
```

```python
%%writefile app.py
# streamlit 패키지 추가
import streamlit as st

st.subheader('Badges', divider='gray')
# Badge 추가
# 배경색이 칠해진 텍스트 생성(기본 blue)
st.badge('New')
# color=색상으로 색상 변경 가능 (색상은 blue, green orange, red, violet, gray/grey 가능)
st.badge('Deprecated', color='grey')
# icon=':material/아이콘이름:'으로 아이콘 지정 가능
# 아이콘 검색: https://fonts.google.com/icons?icon.set=Material+Symbols&icon.style=Rounded
st.badge('Success', color='green', icon=':material/check:')
# 텍스트 옆에 badge 추가
# st.markdown 이용
st.markdown('성공! :green-badge[:material/check: Success]')

st.subheader('Caption', divider='gray')
# Caption 추가
st.caption('나는 :green[Caption]입니다.')
```

```python
%%writefile app.py
# streamlit 패키지 추가
import streamlit as st

# 버튼
clicked = st.button('button')
# 버튼 클릭 시 메시지 출력
if clicked:
    st.write('버튼이 클릭되었습니다.')

# 체크박스
checked1 = st.checkbox('checkbox1')
checked2 = st.checkbox('checkbox2')
# 체크박스 체크 시 메시시 출력
if checked1:
    st.write('checkbox1이 선택되었습니다.')
if checked2:
    st.write('checkbox2가 선택되었습니다.')
```

```python
%%writefile app.py
# streamlit 패키지 추가
import streamlit as st

# 라디오 버튼
selected = st.radio(
    'Radio Button', # 라디오 버튼 제목
    [':rainbow[Radio1]', '***Radio2***', 'Radio3 :material/check:'], # 라디오 버튼 옵션
    captions=[  # 각 버튼 캡션 지정
        '라디오 버튼 1번 옵션',
        '라디오 버튼 2번 옵션',
        '라디오 버튼 3번 옵션'],
    index=None, # 기본 옵션 인덱스 지정(0~n-1 또는 None)
)
# 라디오 버튼 선택 시 메시지 출력
if selected:
    st.write(f'선택된 옵션: {selected}')
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd

dataframe = pd.DataFrame({'first column': [1, 2, 3, 4],
                          'second column': [10, 20, 30, 40]})
# 다운로드 버튼 연결
# 데이터프레임 다운로드
@st.cache_data # 동일 파일인 경우 캐시 동작
def to_csv(df):
    return df.to_csv(index=False).encode('utf-8')

csv = to_csv(dataframe)
st.download_button(
    label='CSV로 다운로드',data=csv, file_name='sample.csv', mime='text/csv')
```

```python
%%writefile app.py
import streamlit as st
import time

st.header('프로그레스 바')
progress_text = "작업을 진행 중입니다. 잠시만 기다려주세요"
my_bar = st.progress(0, text=progress_text)

for percent_complete in range(100):
    time.sleep(0.01)
    my_bar.progress(percent_complete + 1, text=progress_text+('.'*(percent_complete%4)))
else:
    my_bar.progress(100, text='작업이 완료되었습니다.')
```

```python
%%writefile app.py
import streamlit as st

if 'button_state' not in st.session_state:
    st.session_state['button_state'] = False
if 'button_write' not in st.session_state:
    st.session_state['button_write'] = ''

# 버튼 클릭
left, center, right = st.columns(3)
button = center.button('버튼을 눌러보세요')

if button:
    msg = '버튼이 눌렸습니다.'
    st.write(msg)
    st.session_state['button_state'] = True
    st.session_state['button_write'] = msg
else:
    if st.session_state['button_state'] == True:
        st.write(st.session_state['button_write'])

# 체크 박스
agree = st.checkbox('동의 하십니까?')

if agree:
    st.write('동의 해주셔서 감사합니다 :100:')
```
